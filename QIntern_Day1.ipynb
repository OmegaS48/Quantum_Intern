```python
from qiskit import QuantumCircuit
```

```python
q=QuantumCircuit(3,3)
```

```python
q.cx(0,1)
```

```python
q.cx(0,2)
```

```python
q.draw(output="mpl")
```

```python
for i in range(3):
    q.h(i)
```

```python
q.draw(output="mpl")
```

```python
q.measure(0,0)
```

```python
q.draw(output="mpl")
```

```python
from qiskit_aer import AerSimulator
```

```python
sim=AerSimulator()
```

```python
result=sim.run(q,shots=10000).result()
```

```python
# 3 qubits + 3 classical bits
qc = QuantumCircuit(3, 3)

# Superposition
qc.h(0)

# Entanglement
qc.cx(0, 1)
qc.cx(0, 2)

# Measurement
qc.measure(0, 0)
qc.measure(1, 1)
qc.measure(2, 2)
```

```python
result=sim.run(q,shots=1000).result()
```

```python
count=result.get_counts()
```
