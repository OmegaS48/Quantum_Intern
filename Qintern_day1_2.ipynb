```python
from qiskit import QuantumCircuit
```

```python
q=QuantumCircuit(3,3)
```

```python
q.h(0)
```

```python
q.cx(0,1)
```

```python
q.cx(0,2)
```

```python
q.draw()
```

```python
q.measure(0,0)
q.measure(1,1)
q.measure(2,2)
```

```python
from qiskit_aer import AerSimulator

simulator = AerSimulator()

result = simulator.run(q, shots=5000).result()
counts = result.get_counts()

print(counts)
```
