```python
import numpy as np
```

```python
from qiskit import QuantumCircuit
```

```python
a=1/np.sqrt(2)
b=1/np.sqrt(2)
```

```python
q=QuantumCircuit(5,2)
```

```python
q.initialize([a,b],0)
```

```python
q.cx(0,1)
```

```python
q.cx(0,2)
```

```python
q.x(0) #manually added
```

```python
q.draw(output="mpl")
```

```python
q.cx(0,3)
```

```python
q.cx(1,3)
```

```python
q.cx(1,4)
```

```python
q.cx(2,4)
```

```python
q.draw(output='mpl')
```

```python
q.measure(3,0)
q.draw(output="mpl")
```

```python
q.measure(4,1)
q.draw(output="mpl")
```

```python
from qiskit_aer import AerSimulator
sim=AerSimulator()
```

```python
job = sim.run(q, shots=100)
```

```python
result=job.result()
```

```python
count=result.get_counts()
```
