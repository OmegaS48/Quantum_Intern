```python
from qiskit import QuantumCircuit
```

```python
q=QuantumCircuit(3,3)
```

```python
q.draw("mpl")
```

```python
#superposition
q.h(0)
q.h(1)
q.h(2)
```

```python
q.draw("mpl")
```

```python
q.measure(0,0)
q.measure(1,1)
q.measure(2,2)
```

```python
from qiskit_aer import AerSimulator
sim=AerSimulator()
job = sim.run(q,shots=1000).result()
count=job.get_counts()
print(count)
```

```python
import matplotlib.pyplot as plt
```

```python
labels = list(count.keys())
values = list(count.values())

fig, ax = plt.subplots()
ax.bar(labels, values, color="#0affe0")
ax.set_xlabel("Bitstring")
ax.set_ylabel("Results")
ax.set_title("Quantum circuit output")
plt.tight_layout()
plt.show()
```

```python
for shots in [10, 100, 1000, 10000]:
 
    qc = QuantumCircuit(1, 1)
    qc.h(0)
    qc.measure(0, 0)
 
    result = sim.run(qc, shots=shots).result()
    counts = result.get_counts()
 
    print("Shots:", shots)
    print(counts)
labels = list(counts.keys())
values = list(counts.values())

fig, ax = plt.subplots()
ax.bar(labels, values, color="#64b122")
ax.set_xlabel("Bitstring")
ax.set_ylabel("Results")
ax.set_title("Quantum circuit output")
plt.tight_layout()
plt.show()
```
