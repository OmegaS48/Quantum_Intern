```python
a=1/np.sqrt(2)
b=1/np.sqrt(2)
```

```python
from qiskit import QuantumCircuit
```

```python
q=QuantumCircuit(5,2)
```

```python
q.initialize([a,b],0)
```

```python
q.cx(0,1)
q.cx(0,2)
q.h(0)
q.h(1)
q.h(2)
```

```python
q.cx(3,0)
```

```python
q.cx(3,1)
```

```python
q.cx(4,1)
```

```python
q.cx(4,2)
```

```python
q.h(3)
q.h(4)
```

```python
q.measure(3,0)
```

```python
q.measure(4,1)
```

```python
q.draw(output="mpl")
```
